```julia
# Inclination for the demonstration
i_demo = deg2rad(98.1)

# Design the repeat orbit
a_repeat = design_repeat_orbit(
    409,
    28,
    i_demo
)

# Conventional Orbit (does not satisfy repeat condition)
conv_sat = Satellite(
    "Conventional Orbit",
    a_repeat + 50.0,      # Slightly different semi-major axis
    0.0,                  # Eccentricity
    i_demo,               # Inclination
    deg2rad(0.0),         # RAAN
    deg2rad(0.0),         # Argument of Perigee
    deg2rad(0.0)          # Mean Anomaly
)

# Designed Repeat Orbit
repeat_sat = Satellite(
    "Repeat Orbit",
    a_repeat,
    0.0,
    i_demo,
    deg2rad(0.0),
    deg2rad(0.0),
    deg2rad(0.0)
)
```

```julia
repeat_result = propagate_j2(
    repeat_sat,
    GM,
    ωE;
    dt=10.0,
    revolutions=20
)

conv_result = propagate_j2(
    conv_sat,
    GM,
    ωE;
    dt=10.0,
    revolutions=20
)
```

```julia
p1 = plot_groundtrack(
    conv_result.latitudes,
    conv_result.longitudes;
    title="Conventional Sampling"
)
savefig(p1,"figures/conventional_sampling.png")

p2 = plot_groundtrack(
    repeat_result.latitudes,
    repeat_result.longitudes;
    title="Repeat-Orbit Sampling"
)

savefig(p2,"figures/repeat_sampling.png")

p = plot(
    p1,
    p2,
    layout=(1,2),
    size=(1400,500)
)

display(p)

savefig(
    p,
    "figures/repeat_vs_conventional_sampling.png"
)
```
